# Q1. What is Machine Learning? Explain the main types of Machine Learning with examples.

- Answer

Machine Learning is a technique where computers learn automatically from data and improve their performance with experience, without being explicitly programmed for every task.

- Types of Machine Learning


1. **Supervised Learning**
Concept: Model is trained on labeled data (input + correct output).

Examples:

Classification: Spam email detection (spam / not spam).

Regression: Predicting house prices based on size and location.

Applications: Fraud detection, medical diagnosis, weather forecasting.



2. **Unsupervised Learning**
Concept: Model works with unlabeled data to find hidden patterns or groupings.

Examples:

Clustering: Customer segmentation in marketing.

Association: Market basket analysis (items often bought together).

Applications: Recommendation systems, anomaly detection, social network analysis.

3. **Reinforcement Learning**
Concept: Model learns by interacting with an environment and receiving rewards or penalties.

Examples:

Training robots to walk.

Game-playing AI (e.g., AlphaGo).

Applications: Self-driving cars, robotics, personalized recommendations


# Q2. What is Supervised Learning? Explain the difference between Regression and Classification.

 - Answer

Supervised Learning is a type of Machine Learning where the model is trained using labeled data. This means that for each input, the correct output (label) is already known. The algorithm learns the mapping between inputs and outputs, and then can predict outcomes for new, unseen data.

Example: Training a spam filter. Emails are labeled as spam or not spam. The model learns patterns and predicts whether a new email is spam.


Step 1: Identify the type of output

Regression: Produces a continuous numerical value.

Classification: Produces a discrete category or label.


Step 2: Define the goal

Regression: The goal is to predict a quantity (e.g., price, temperature, age).

Classification: The goal is to assign inputs into groups (e.g., spam vs. not spam, disease vs. no disease).


Step 3: Give examples

Regression Example: Predicting house prices based on size and location.

Classification Example: Predicting whether an email is spam or not.



Step 4: Algorithms used

Regression Algorithms: Linear Regression, Support Vector Regression, Decision Trees (for regression).

Classification Algorithms: Logistic Regression, Random Forest, Neural Networks, Decision Trees (for classification).



Step 5: Evaluation metrics

Regression Metrics: Mean Squared Error (MSE), Root Mean Squared Error (RMSE).

Classification Metrics: Accuracy, Precision, Recall, F1-score.

# Q5. Explain the Machine Learning workflow steps from data collection to model evaluation.

- Answer

The process of building a Machine Learning model follows a structured workflow. Here are the main steps from data collection to model evaluation:

Step 1: Data Collection
Gather raw data from sources like databases, sensors, surveys, or web scraping.

Example: Collecting customer purchase records for sales prediction.

Step 2: Data Preprocessing
Clean the data (remove duplicates, handle missing values).

Transform data (normalize, scale, encode categorical variables).

Example: Converting “City” names into numerical codes for ML algorithms.


Step 3: Feature Selection & Engineering
Select important variables (features) that influence the target.

Create new features if needed (e.g., “Age Group” from “Age”).

Example: In house price prediction, features could be size, location, and number of rooms.


Step 4: Splitting the Dataset
Divide data into training set (to train the model) and testing set (to evaluate performance).

Common split: 70% training, 30% testing.


Step 5: Model Selection & Training
Choose an appropriate algorithm (Linear Regression, Decision Trees, Neural Networks, etc.).

Train the model using the training dataset.

Example: Using Linear Regression to predict house prices.

Step 6: Model Evaluation
Test the model on unseen data (testing set).

Use evaluation metrics depending on the task:

Regression → MSE, RMSE, R²

Classification → Accuracy, Precision, Recall, F1-score

Step 7: Model Deployment (Optional)
Deploy the trained model into real-world applications (web apps, mobile apps, business systems).

Example: A recommendation system on an e-commerce website.


Step 8: Monitoring & Maintenance
Continuously monitor performance.

Update/retrain the model when new data arrives or patterns change.

# Q10. Train a Linear Regression model on the training data.

Q10. Train a Linear Regression model and perform the following:

Make predictions

Calculate MAE

Calculate MSE

Calculate RMSE

Calculate R² Score

Calculate Adjusted R²

Write the Linear Regression equation

Interpret the model coefficients

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load dataset
df = pd.read_csv("BostonHousing.csv")

# Separate independent variables (X) and dependent variable (y)
X = df.drop(columns=["medv"])
y = df["medv"]

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# Create and train Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate evaluation metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

# Calculate Adjusted R²
n = X_test.shape[0]     # Number of observations
p = X_test.shape[1]     # Number of predictors

adjusted_r2 = 1 - ((1 - r2) * (n - 1) / (n - p - 1))

# Display results
print("Model Evaluation Metrics")
print("-" * 35)
print(f"MAE         : {mae:.4f}")
print(f"MSE         : {mse:.4f}")
print(f"RMSE        : {rmse:.4f}")
print(f"R² Score    : {r2:.4f}")
print(f"Adjusted R² : {adjusted_r2:.4f}")

# Display coefficients
coefficients = pd.Series(model.coef_, index=X.columns)

print("\nModel Coefficients:")
display(coefficients.to_frame(name="Coefficient"))

# Display predictions
predictions = pd.DataFrame({
    "Actual MEDV": y_test.values,
    "Predicted MEDV": y_pred
})

print("\nActual vs Predicted Values:")
display(predictions.head(10))

# Create regression equation
equation = f"medv = {model.intercept_:.4f}"

for feature, coefficient in coefficients.items():
    sign = "+" if coefficient >= 0 else "-"
    equation += f" {sign} {abs(coefficient):.4f}({feature})"

print("\nLinear Regression Equation:")
print(equation)
```
